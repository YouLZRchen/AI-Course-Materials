先给出以下问题的解答过程和再给出答案：甲、乙、丙，丁四人各持有一张卡片，卡片数字为3,5,7,9，每人一张互不重复。已知四条线索：甲的数字不是最小数，乙的数字大于丙，丙的数字不等于5，丁不是7也不是9。请推理出四人分别对应的数字。


```python
import ollama

def chat_with_ollama():
    # 在这里换成你要使用的模型名称
    MODEL_NAME = "deepseek-r1:1.5b-qwen-distill-fp16"
    
    print("=== Ollama 连续对话助手 ===")
    print(f"当前模型：{MODEL_NAME}")
    print("输入 'quit' / 'exit' 退出对话\n")

    # 保存对话历史，实现上下文连续问答
    messages = []

    while True:
        # 终端输入问题
        user_input = input("你：").strip()
        
        # 退出条件
        if user_input.lower() in ["quit", "exit"]:
            print("对话结束！")
            break
        if not user_input:
            print("请输入有效内容~")
            continue

        # 把用户问题加入历史
        messages.append({"role": "user", "content": user_input})

        try:
            # 调用 ollama 生成回答（流式输出）
            print("\nAI：", end="", flush=True)
            response = ollama.chat(model=MODEL_NAME, messages=messages, stream=True)
            
            # 逐字打印输出
            ai_reply = ""
            for chunk in response:
                content = chunk['message']['content']
                ai_reply += content
                print(content, end="", flush=True)
            
            print("\n")
            # 把 AI 回答也加入历史，保证下一轮有上下文
            messages.append({"role": "assistant", "content": ai_reply})

        except Exception as e:
            print(f"\n出错啦：{str(e)}\n")
            # 出错后移除刚才的用户消息，避免历史混乱
            messages.pop()

if __name__ == "__main__":
    chat_with_ollama()
```
